## Exercise 2: Average Treatment Effect on the Treated
### Question 1
The average treatment effect on the treated (ATT) refers to the treatment effect for those who actually received treatment. It gives the average over the subpopulation of treated people of the treatment effect. On the other hand, the average treatment effect (ATE) is the average effect of the treatment across the entire eligible sample. The ATE represents a weighted average of the ATT and the treatment effect for those who did not receive the treatment.

### Question 2
Under consistency assumption, we have:
$$E[Y_i|T_i=1] = E[Y_i(1)|T_i=1]$$
$$E[Y_i|T_i=0] = E[Y_i(0)|T_i=0]$$
Under weak ignorability assumption holding for the control outcome, we have:
$$E[Y_i(0)|T_i=1] = E[Y_i(0)|T_i=0]$$

Then for ATT
\begin{equation*}
    \begin{split}
        \tau^t & = E[Y_i(1) - Y_i(0)|T_i = 1]\\
        & = E[Y_i(1)|T_i=1] - E[Y_i(0)|T_i=1] \;\;\;\;\mbox{linearity of the expectation}\\
        & = E[Y_i(1)|T_i=1] - E[Y_i(0)|T_i=0] \;\;\;\;\mbox{weak ignorability}\\
        & = E[Y_i|T_i=1] - E[Y_i|T_i=0] \;\;\;\;\;\;\;\;\;\;\;\;\mbox{consistency/SUTVA}
    \end{split}
\end{equation*}

## Exercise 3: Bernoulli trial
### Question 1
**1. What distribution does $T_i$ follow?**

**Answer:** The treatment Variable $T_i$ follows the Bernoulli Distribution.
    
**2. What is $E[N_t]$?**

**Answer:** We know that $N_t = \sum_{i=1}^n T_i$
    
$E[N_t] = \sum_{i=1}^n E[T_i]$
    
Let us first solve for the expected value of $T_i$. $T_i$ is a random variable with $\Pr(T_i = 1) = p$ and $\Pr(T_i = 0) = (1-p)$. 
    
 So, $E[T_i] = p*1 + (1-p)*0 = p$ 
    
Returning to our main answer,
    
$E[N_t] = \sum_{i=1}^n p$ 
    
$E[N_t] = n*p$ (Adding up $p$ $n$ times in the summation)
    
That implies, $E[N_t] = n*p = N_t$ (total number of units multiplied by the probability of treatment is nothing but total number of treated units!) 


**3.What is $Var[N_t]$?**
    
**Answer:** Again let us first compute the Variance of $T_i$. 
    
$Var[T_i] = E[T_i^2] - E[T_i]^2 $ 
    
$E[T_i^2] = 1^2*p + 0^2*(1-p) = p$ (Same as $E[T_i]$ as computed in previous question.) 
    
So, $Var[T_i] = p - p^2 = p(1-p)$ 
    
Now, $Var[N_t] = \sum_{i=1}^n Var(T_i) + \sum_{j \neq i}^n Cov(T_i, T_j)$ 
    
Since, the assignment of treatment is iid, all the covariance terms will be 0. 
    
So, $Var[N_t] = \sum_{i=1}^n Var(T_i)$ 
    
$= \sum_{i=1}^n p(1-p)$ 
    
$Var(N_t) = np(1-p)$ 
    
Therefore from the expected value and variance, it is evident that $N_t$ follows Binomial Distribution. (Sum of Bernoulli distributed random variables.) 
    
**4. Suppose that we wanted the \textbf{expected} number of treated units in our Bernoulli trial to be the same as the number of treated units as a completely randomized experiment with $n_t$ treated units. What value of $p$ should we choose?**
     
**Answer:** We want the number of treated units under Bernoulli trail to be same as the number of treated units in a completely randomized experiment with $n_t$ treated units. For this to happen the Expected value of $N_t$ should be equal to $n_t$.  
     
So, $E[N_t] = n_t$ 
     
This implies, $n*p = n_t$
     
$p = \frac{n_t}{n}$.


### Question 2

$E[\hat{\tau}_{IPW}] = \frac{1}{n}\sum_{i=1}^n E[Y_i\frac{T_i}{p}] - E[Y_i\frac{1-T_i}{1-p}]\;\;\;\;\;\;\;\;\mbox{linearity of expectation}$ \\

$E[\hat{\tau}_{IPW}] = \frac{1}{n}\sum_{i=1}^n E[\frac{T_iY_i(1)}{p}] -  \frac{1}{n}\sum_{i=1}^n E[\frac{(1-T_i)Y_i(0)}{1-p}]\;\;\;\;\;\;\;\;\mbox{consistency and SUTVA}$ 

$E[\hat{\tau}_{IPW}] = \frac{1}{n}\sum_{i=1}^n \frac{E[T_iY_i(1)]}{p} -  \frac{1}{n}\sum_{i=1}^n \frac{E[(1-T_i)Y_i(0)]}{1-p}\;\;\;\;\;\;\;\;\mbox{linearity of expectation}$ 

$E[\hat{\tau}_{IPW}] = \frac{1}{n}\sum_{i=1}^n \frac{E[T_i]E[Y_i(1)]}{p} -  \frac{1}{n}\sum_{i=1}^n \frac{E[(1-T_i)]E[Y_i(0)]}{1-p}\;\;\;\;\;\;\;\;\mbox{ignorability and expectation of 2 independent variables}$ 



$E[\hat{\tau}_{IPW}] = \frac{1}{n}\sum_{i=1}^n E[Y_i(1)]\frac{p}{p} -  \frac{1}{n}\sum_{i=1}^n E[Y_i(0)]\frac{1-p}{1-p}\;\;\;\;\;\;\;\;\mbox{expectation of a Bernoulli random variable}$ \\

$E[\hat{\tau}_{IPW}] = \frac{1}{n}\sum_{i=1}^n E[Y_i(1)] -  \frac{1}{n}\sum_{i=1}^n E[Y_i(0)]$ \\

$E[\hat{\tau}_{IPW}] = \frac{1}{n}\sum_{i=1}^n E[Y_i(1) - Y_i(0)]$ \\

$E[\hat{\tau}_{IPW}] = \tau$ \\

Therefore, we can say that $\hat{\tau}_{IPW}$ is an unbiased estimator of $\tau$.

```python
import pandas as pd
import numpy as np
import scipy.stats as sps
import matplotlib.pyplot as plt
```

### Question 1

```python
gotv = pd.read_excel('gotv_individual.xlsx')
```

```python
gotv = gotv.assign(sex_binary=gotv.sex.map({'male': 0, 'female': 1}))
```

```python
gotv = gotv.assign(age=2006-gotv.yob)
```

```python
num_var = ['hh_size', 'g2000', 'g2002', 'g2004', 'p2000', 'p2002', 'p2004', 'voted', 'sex_binary', 'age']
cat_var = ['treatment']
transformations = {v: 'mean' for v in num_var}
transformations.update({v: pd.Series.mode for v in cat_var})
gotv_house = gotv.groupby('hh_id').agg(transformations)
gotv_house.head()
```

The authors work at the household level, because the individual level poses some issues:
1. For individual level, we can't determine who reads the mailer. 
2. Household member may affect each other.

### Question 3

```python
potential_treatments = list(gotv_house.treatment.unique())
potential_treatments.remove(' Control')
res_dict = {}
for t in potential_treatments:
    df = gotv_house[gotv_house.treatment==t]
    res_dict[t] = df.voted.mean() - df.voted.mean()
```

```python
res_df = pd.DataFrame.from_dict(res_dict, orient='index', columns=['ate'])
res_df
```

We are assuming SUTVA (same treatment version for each treatment, and no interaction between households), as well as ignorability and positivity from the randomization.

### Question 4

```python
control = gotv_house[gotv_house.treatment==' Control']
neighbors = gotv_house[gotv_house.treatment==' Neighbors']

var_ate = neighbors.voted.var() / len(neighbors) + control.voted.var() / len(control)

tau = neighbors.voted.mean() - control.voted.mean()
print(tau)

Zn = tau / np.sqrt(var_ate)
pval = sps.norm.sf(np.abs(Zn))*2

print('statistic', Zn, 'pvalue', pval)
```

The p-value is very small

```python
gotv_neighbors_control = gotv_house[gotv_house.treatment.isin([' Control', ' Neighbors'])]
N = len(gotv_neighbors_control)
rg = np.random.default_rng(125)
N_iter = 1000
all_Zs = np.empty(N_iter)
for i in range(N_iter):
    # assigns treatment to N units, each unit has the same prob of receiving treatment
    gotv_neighbors_control = gotv_neighbors_control.assign(
      T_permute = rg.permutation(gotv_neighbors_control.treatment))
    perm_gotv_1 = gotv_neighbors_control[gotv_neighbors_control.T_permute==' Neighbors'].voted
    perm_gotv_0 = gotv_neighbors_control[gotv_neighbors_control.T_permute==' Control'].voted
    all_Zs[i] = (perm_gotv_1.mean() - perm_gotv_0.mean())/np.sqrt(perm_gotv_1.var()/len(perm_gotv_1) + perm_gotv_0.var()/len(perm_gotv_0))
```
